```python
df['Age'] = df['Age'].fillna(df['Age'].mean()).astype('int64')
df.info()
```

```python
df["Cabin"] = df["Cabin"].fillna("Unknown")
df
```

```python
df['Embarked'] = df['Embarked'].fillna("Unknown")
```

```python
df['Fare'] = df['Fare'].fillna(df['Fare'].mean()).astype('int64')
df
```

```python
df["Ticket"] = df["Ticket"].str.replace(r"[A-Za-z/.\s]", "", regex=True)
df
```

```python
df['Survived'] = df['Survived'].replace({
    0: 'died',
    1: 'survived'
}).astype('object')
df
```

```python
# "Unknown" olan sətrləri silirik
df = df[df["Embarked"] != "Unknown"]
df
```

```python
df.info()
```

```python
# Visualization
```

```python
df['Survived'].value_counts().plot(kind='bar',color='g')
plt.xticks(rotation=0)
plt.show()
```

```python
sns.countplot(x="Sex", hue="Survived", data=df,palette='Set2')
plt.show()
```

```python
df["Age"].hist(bins=30)
plt.show()
```

```python
df['Survived'].value_counts().plot.pie(autopct='%1.1f%%', colors=['red','green'])
plt.figure(figsize=(10,10))
plt.show()
```

```python
numeric_cols = ["Age", "Fare", "SibSp", "Parch"]
sns.pairplot(df[numeric_cols + ["Survived"]], hue="Survived", diag_kind="hist")
plt.show()
```

```python
pclass_survived = df.groupby("Pclass")["Survived"].value_counts().unstack().fillna(0)

pclass_survived.plot(kind="area", stacked=True, color=["red","green"], alpha=0.6)
plt.xlabel("Pclass")
plt.ylabel("Count")
plt.title("Stacked Area Plot of Survived by Pclass")
plt.show()
```

```python
corr = df[['Pclass','Fare','Age']].corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('heatmap')
plt.show()
```

```python
sns.histplot(data=df,x="Sex",hue="Survived",multiple="stack",palette={"died":"red","survived":"green"})
plt.xlabel("Sex")
plt.ylabel("Count")
plt.title("Stacked Bar Plot of Survival by Sex")
plt.show()
```

```python
df['Embarked'].value_counts().plot(kind='barh', color='purple')
plt.title('Embarked Ports')
plt.xlabel('Number of Passengers')
plt.ylabel('Embarked')
plt.show()
```
